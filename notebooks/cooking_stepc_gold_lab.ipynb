## Progress overview

````python
editor.summary()
editor.show_overview(frame='COOKING_CREATION')
````

## Label rows

1. Read the **snippet** (target highlighted).
2. Set **homonym_check** (`food_sense` / `other_sense` / `metaphor`).
3. Fill **Step A** and **Step B**.
4. Fill **Step C qualia** only when stated in the snippet (leave blank otherwise).
5. Check **labelled**, click **Save & next** — pending queue, auto-saves CSV.
6. Jump rows via the **dropdown** or **Prev/Next** (after scanning the overview).


> **N.B.** The overview table is **browse-only** — not Excel-style inline edit.
> Edits go through the row editor below.
>
> With `pending_only=True`, finished (`labelled=true`) rows drop out of the queue,
> so you cannot reopen them. To fix a mistake (e.g. leftover `mede`):
> set `pending_only=False`, re-run the cell, pick the row in the dropdown.
> For dropped non-food rows also set **Advanced → frame → NONE**.


````python
# pending_only=True = unfinished queue only. False = revisit labelled rows.
labeller = RowLabeller(editor, frame_filter='COOKING_CREATION', pending_only=True)
labeller.display()
````
