## Setup

````python
# Dev deps (ipykernel, ipywidgets, itables) install with: uv sync
from pathlib import Path

from data_io import resolve
from trifecta_annotation.gold_nb import GoldBatchEditor, RowLabeller

SCRATCH = Path(resolve('trifecta_gold')).parent
BATCH_PATH = SCRATCH / 'eval' / 'cooking_stepc_batch.csv'

editor = GoldBatchEditor(BATCH_PATH)
editor.summary()
````

## Export batch (if missing)

Run once in the terminal:

```bash
uv run python scripts/export_clear_frame_examples.py --summary --pool-summary \
  --frames COOKING_CREATION \
  --frame-quota "COOKING_CREATION:30" --limit 30 \
  --output-path "$SCRATCH/eval/cooking_stepc_batch.csv"
```


## Progress overview

````python
editor.summary()
editor.show_overview(frame='COOKING_CREATION')
````

## Label rows

1. Read the **snippet** (target highlighted).
2. Set **homonym_check** (`food_sense` / `other_sense` / `metaphor`).
3. Fill **Step A** and **Step B**.
4. Fill **Step C qualia** only when stated in the snippet (leave blank otherwise).
5. Check **labelled**, click **Save & next** — pending queue, auto-saves CSV.
6. Overview table shows **snippets** + short qualia columns; dropdown to jump.


````python
labeller = RowLabeller(editor, frame_filter='COOKING_CREATION', pending_only=True)
labeller.display()
````

## Save to disk (optional)

````python
# Autosave is on by default — each Save / Save & next writes the batch CSV.
# Run this only to confirm or if you passed autosave=False:
saved = editor.save()
print('Wrote', saved)
editor.summary()
````
